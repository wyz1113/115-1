```python
def h(n):
  if n==1:
    return 1
  return n*h(n-1)

h(5)
```

```python
def h(n):
  if n==1:
    return 1
  if n==2:
    return 2
  return h(n-1)+h(n-2)
sum=h(10)
print(f"總共有{sum}種方式")
```
